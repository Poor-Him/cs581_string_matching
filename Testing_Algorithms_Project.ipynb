```python
import tempfile, os

base = "/content/drive/MyDrive/2026 Algorithms Project"
texts_dir = f"{base}/Texts"

# Load patterns
'''with open(f"{base}/patterns/dna_pattern_1.txt") as f:
    dna_pat_1 = "\\n".join(line.strip() for line in f.read().splitlines())
with open(f"{base}/patterns/dna_pattern_2.txt") as f:
    dna_pat_2 = "\\n".join(line.strip() for line in f.read().splitlines())'''
with open(f"{base}/patterns/dna_pattern_1.txt") as f:
    dna_pat_1 = f.read().strip().replace("\n", "")
with open(f"{base}/patterns/dna_pattern_2.txt") as f:
    dna_pat_2 = f.read().strip().replace("\n", "")
with open(f"{base}/patterns/text_pattern_1.txt") as f:
    text_pat_1 = "\\n".join(line.strip() for line in f.read().splitlines())
with open(f"{base}/patterns/text_pattern_2.txt") as f:
    text_pat_2 = "\\n".join(line.strip() for line in f.read().splitlines())

script = f"{base}/Programs/Rabin-Karp.py"

def run_script(script, filename, pattern, repeat=1, ignore_case=False, d=0, q=0):
    with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False, encoding='utf-8') as f:
        f.write(pattern)
        tmp = f.name
    try:
        flags = "--ignore-case" if ignore_case else ""
        if(d != 0 and q != 0):
            !python "{script}" --input-file "{filename}" --pattern-file "{tmp}" --repeat {repeat} {flags} --d {d} --q {q}
        else:
            !python "{script}" --input-file "{filename}" --pattern-file "{tmp}" --repeat {repeat} {flags}
    finally:
        print(f"d: {d}, q: {q}\n")
        os.unlink(tmp)
if("DFA" in script):
    for filename in os.listdir(texts_dir):
        full_path = os.path.join(texts_dir, filename)
        if filename.endswith(".fna"):
            for pat in [dna_pat_1, dna_pat_2]:
                run_script(script, full_path, pat, ignore_case=True)
                print("")
        elif filename.endswith(".txt"):
            for pat in [text_pat_1, text_pat_2]:
                run_script(script, full_path, pat)
                print("")
else:
    for filename in os.listdir(texts_dir):
        if filename.endswith(".fna"):
            for pat in [dna_pat_1, dna_pat_2]:
                if("Rabin" in script):
                    run_script(script, filename, pat, ignore_case=True, d=4, q=1000000007)
                else:
                    run_script(script, filename, pat, ignore_case=True)
                print("")
        elif filename.endswith(".txt"):
            for pat in [text_pat_1, text_pat_2]:
                if("Rabin" in script):
                    run_script(script, filename, pat, d=512, q=1000000007)
                else:
                    run_script(script, filename, pat)
                print("")
```
